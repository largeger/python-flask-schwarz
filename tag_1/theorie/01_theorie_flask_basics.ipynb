## 1. Was ist Flask?
Flask ist ein **Micro-Framework**. Das bedeutet, es ist sehr leichtgewichtig und gibt dir viel Freiheit, wie du deine Anwendung strukturierst.

```python
# Wir können Flask hier nicht direkt 'laufen' lassen wie eine App,
# aber wir können den Code vorbereiten.

from flask import Flask

app = Flask(__name__)

@app.route("/")
def home():
    return "Willkommen auf der Homepage!"

print("Flask App Instanz erstellt!")
```

## 2. Der Request-Response Zyklus
1. **Client** (Browser) sendet einen **Request** (Anfrage) an eine URL.
2. **Server** (Flask) empfängt den Request und schaut, welche Funktion für diese URL zuständig ist (**Routing**).
3. Die Funktion wird ausgeführt und liefert eine **Response** (Antwort) zurück.

## 3. Dynamisches Routing
Wir können Teile der URL als Variablen nutzen.

```python
from threading import Thread

app = Flask(__name__)

@app.route("/hello/<name>")
def say_hello(name):
    return f"Hallo {name.capitalize()}!"

# Typprüfung in der URL
@app.route("/calc/square/<int:number>")
def square(number):
    return f"Das Quadrat von {number} ist {number**2}"

def run_app():
    app.run(host="127.0.0.1", port=5000, debug=True, use_reloader=False)

thread = Thread(target=run_app)
thread.start()
```

## 4. JSON APIs
Moderne Backends liefern oft kein HTML, sondern Daten im JSON-Format.

```python
from threading import Thread

app = Flask(__name__)

@app.route("/api/info")
def api_info():
    data = {
        "framework": "Flask",
        "version": "3.0",
        "type": "Backend"
    }
    return data # Flask macht automatisch JSON daraus

def run_app():
    app.run(host="127.0.0.1", port=5000, debug=True, use_reloader=False)

thread = Thread(target=run_app)
thread.start()
```
